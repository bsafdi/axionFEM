# dc SQUID readout chains

In lumped-element experiments, a dc SQUID converts flux from a tuned axion pickup into an electrical output. Its measurement noise is correlated with circulating-current noise that feeds back into the pickup. An array and preamplifier add further noise. In this tutorial we follow the chain through resonator coupling and scan reach.

> **Beta.** The `axionfem.squid` chain models are beta in this release: they implement the Tesche-Clarke noise model of Ankel et al., arXiv:2504.20398, with its rounded response functions and a second stage scaled from one measured prototype, and their interfaces may still change.

## Noise parameters

For one-sided per-Hz power spectral densities,

$$
\eta(\omega)=
\frac{\sqrt{S^{(1)}_{\Phi,\rm tot}S^{(1)}_J-(S^{(1)}_{\Phi J})^2}}{\hbar},
\qquad
\epsilon_{\rm uc}(\omega)=\frac{S^{(1)}_{\Phi,\rm tot}}{2L_{\rm sq}}.
$$

The spectra describe total flux noise, first-stage circulating-current noise, and their real cross-spectrum. $L_{\rm sq}$ is the first-stage loop inductance; $\omega$ is angular frequency. Dimensionless $\eta$ accounts for correlation; $\epsilon_{\rm uc}$ is uncoupled energy sensitivity in joule-seconds. Smaller $\eta$ improves scan sensitivity at fixed detector and thermal noise.

## Custom devices

We set junction and loop parameters, then attach the later stages. `validity()` reports junction and thermal diagnostics; the Tesche–Clarke model assumes weak coupling and optimal bias. It omits $1/f$ and quantum noise, later-stage backaction, and SQUID input loading.

We evaluate custom chains and a hot-shunt device to compare noise and check operating-regime warnings.

```python
dev = squid.FirstStage(L_sq_H=80e-12, I0_A=12e-6, R_j_ohm=3.0, T_j_K=0.10, C_j_F=3.0e-12, label="my SQUID")
vd = dev.validity()
print(f"beta_L = {vd['beta_L']:.2f}, beta_C = {vd['beta_C']:.2f}, Gamma = {vd['Gamma']:.1e}, "
      f"f_J = {dev.f_josephson_Hz/1e9:.1f} GHz, warnings: {vd['warnings']}")
print(f"eta_TC = {dev.eta_TC:.3f}   eps_uc/hbar = {dev.eps_uc_J_s/HBAR:.3f}")

mine = squid.Chain(dev, squid.SecondStage(48, 3, T2_K=1.0), squid.RT, label="my SQUID + 48x3 + RT")
# the same device through the preset machinery: replace C1 by `first=`
mine_cryo = squid.chain("C1+low+cryo", first=dev)
for f in (1e6, 10e6, 100e6):
    print(f"{f/1e6:6g} MHz  eta = {mine.eta(f):7.3f}   eps_uc/hbar = {mine.eps_uc_over_hbar(f):7.3f}   "
          f"(cryo amp: eta = {mine_cryo.eta(f) if f >= 5e6 else float('nan'):7.3f})")

# A warning example: the same SQUID with hot shunts (Gamma > 0.05 leaves the TC regime)
with warnings.catch_warnings(record=True) as w:
    warnings.simplefilter("always")
    squid.FirstStage(L_sq_H=80e-12, I0_A=1e-6, R_j_ohm=3.0, T_j_K=4.2)
print("hot device ->", [str(x.message)[:110] + "..." for x in w])
```

## Frequency dependence

Later stages increase imprecision at high frequency; RT preamplifiers switch at 50 MHz. Cryogenic curves begin at 5 MHz.

We plot both noise parameters against frequency to compare readout chains.

```python
def curve(ch, f_lo, f_hi, n=300):
    # eta and eps_uc/hbar on a log grid; NaN at 50 MHz to break the line at the RT preamp switch
    f = np.geomspace(f_lo, f_hi, n)
    if f_lo < 50e6 < f_hi:
        f = np.concatenate([f[f < 50e6], [np.nan], f[f >= 50e6]])
    eta, eps = np.full(f.shape, np.nan), np.full(f.shape, np.nan)
    ok = np.isfinite(f)
    eta[ok], eps[ok] = ch.eta(f[ok]), ch.eps_uc_over_hbar(f[ok])
    return f, eta, eps

fig, axes = plt.subplots(1, 2, figsize=(11, 4.2))
c1 = squid.chain("C1")
for bw in ("low", "medium", "high"):
    for amp, (a, b), ls in (("RT", (1e5, 3e8), "-"), ("cryo", (5e6, 3e8), "--")):
        f, eta, eps = curve(squid.chain(f"C1+{bw}+{amp}"), a, b)
        for ax, y in zip(axes, (eta, eps)):
            ax.plot(f / 1e6, y, color=COL[bw], ls=ls, label=f"C1+{bw}+{amp}")
f, eta, eps = curve(mine, 1e5, 3e8)
for ax, y in zip(axes, (eta, eps)):
    ax.plot(f / 1e6, y, color="0.45", lw=1.2, ls="-.", label="my SQUID + 48x3 + RT")
axes[0].axhline(c1.eta(1e6), color=COL["C1"], ls=":", label="C1 alone")
axes[1].axhline(c1.eps_uc_over_hbar(1e6), color=COL["C1"], ls=":", label="C1 alone")
axes[0].axhline(1.0, color=COL["ideal"], lw=1.2, label="SQL ($\\eta=1$)")
axes[0].set_ylabel(r"noise parameter $\eta = k_BT_{\min}/(\hbar\omega/2)$")
axes[1].set_ylabel(r"uncoupled energy sensitivity $\epsilon_{\rm uc}/\hbar$")
for ax in axes:
    ax.set_xscale("log"); ax.set_yscale("log"); ax.set_xlim(0.1, 300)
    ax.set_xlabel("frequency [MHz]")
axes[0].legend(fontsize=7, ncol=2, loc="upper left")
axes[0].set_title("amplifier noise relative to the SQL")
axes[1].set_title("energy sensitivity (sets the optimal coupling)")
fig.tight_layout()
```

## Coupling to a resonator

We specify the loaded quality factor, total inductance and noise temperature. `M_scan_H()` returns the mutual inductance in henries for maximum scan rate. We plot the coupling across frequency to compare this choice with noise matching at resonance.

```python
Q, T = 1e6, 0.03
f = np.geomspace(1e5, 3e7, 200)
fig, ax = plt.subplots(figsize=(6.4, 4.2))
for name, col in (("C1", COL["C1"]), ("C1+low+RT", COL["low"]), ("C1+high+RT", COL["high"])):
    ch = squid.chain(name)
    ax.plot(f / 1e6, ch.kappa_g2_scan(f, Q, T), color=col, label=f"{name}: scan-optimal")
    ax.plot(f / 1e6, ch.on_resonance(f, Q=Q)["kappa_g2_match"], color=col, ls=":",
            label=f"{name}: noise-matched")
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel("frequency [MHz]"); ax.set_ylabel(r"global coupling $\kappa_g^2 = M^2/(L_{\rm sq}L_{\rm tot})$")
ax.set_title(f"Q = {Q:.0e}, T = {T*1e3:.0f} mK")
ax.legend(fontsize=7)
fig.tight_layout()

ch = squid.chain("C1+low+RT")
n1 = ck.n_bose(2 * math.pi * 1e6, T)
print(f"at 1 MHz, 30 mK: n = {n1:.0f}, eta = {ch.eta(1e6):.2f}, xi_opt = {ch.xi_opt(1e6, T):.4f}, "
      f"kappa_g^2 = {ch.kappa_g2_scan(1e6, Q, T):.3e}, M = {ch.M_scan_H(1e6, Q, 1e-6, T)*1e12:.1f} pH for L_tot = 1 uH")
```

## Detector and readout

Mesh convergence should be checked for each geometry.

We solve a toroid for pickup inductance and signal coupling, then rescale the signal coupling for the target magnetic field.

```python
d = af.template("toroid", R=1.0, H=1.0)
res = d.solve(RUNS / "toroid", verbose=0)
B_target = 5.0
PSI = res.Psi * B_target / res.B_max
LP = res.L_p
print(f"solved: Psi = {res.Psi:.3e} H T m^2 at B_max = {res.B_max:.3f} T;  L_p = {LP:.3e} H")
print(f"rescaled to B_max = {B_target} T: Psi = {PSI:.3e} H T m^2")
```

`squid_for_resonator()` supplies a SQUID model at the optimal mutual inductance, holding chain noise at the tuning frequency and converting to the circuit's spectral convention.

We assemble a tuned series readout and attach its SQUID to evaluate the optimally coupled circuit.

```python
f0, Q, T, L_in = 1e6, 1e6, 0.03, 2e-9
ro = ck.series_readout(L_p_H=LP, Psi_SI=PSI, f_Hz=f0, Q=Q, T_K=T, L_in_H=L_in, M_H=1e-9)
eq0 = ro.equivalent(f_Hz=f0)
w0, L_eff, QL, R0 = eq0["omega0_rad_s"], eq0["L_eff_H"], eq0["Q_loaded"], eq0["R_eff_ohm"]
print(f"f0 = {eq0['f0_Hz']:.6g} Hz, L_eff = {L_eff:.4e} H, Q_loaded = {QL:.4e}, R_eff = {R0:.3e} ohm")

chain_p = squid.chain("C1+low+RT", sign=+1)
ro.squid = chain_p.squid_for_resonator(f0, QL, L_eff, T)
sq = ro.squid
print(f"M = {sq.M_H*1e12:.1f} pH;  circuit PSDs: S_Phi = {sq.S_Phi:.3e} Wb^2 s, S_J = {sq.S_J:.3e} A^2 s, "
      f"S_PhiJ = {sq.S_PhiJ:+.3e} Wb A s;  eta_A = {sq.eta_A:.3f}")

eq = ro.equivalent(w0)
print(f"xi = {eq['xi']:.5f} (optimal {chain_p.xi_opt(f0, T):.5f});  eta_A_eff = {eq['eta_A_eff']:.3f}, "
      f"uncorrelated eta = {eq['eta_A_eff_uncorrelated']:.3f};  band shift x0 = {eq['band_shift_x0']:+.1f} half-linewidths")
print("noise-doubling bandwidth = %.1f Hz (resonator linewidth %.2f Hz)"
      % (eq["bandwidth_Hz"]["noise_doubling"], eq["bandwidth_Hz"]["linewidth"]))
```

We plot thermal noise, backaction, imprecision and correlation, all referred to a series voltage in the pickup, to compare their contributions.

```python
x = np.unique(np.concatenate([-np.geomspace(3000, 10, 250), np.linspace(-10, 10, 81), np.geomspace(10, 3000, 250)]))
w = w0 * (1 + x / (2 * QL))

def budget(readout):
    keys = ("N_V_V2_s", "N_V_thermal_V2_s", "N_V_backaction_V2_s", "N_V_imprecision_V2_s", "N_V_correlation_V2_s")
    out = {k: np.empty(w.size) for k in keys}
    for i, wi in enumerate(w):
        dd = readout.noise_psd(float(wi))
        for k in keys:
            out[k][i] = dd[k]
    return out

nb = budget(ro)
fig, ax = plt.subplots(figsize=(7, 4.4))
ax.plot(x, nb["N_V_V2_s"], color="k", lw=2.2, label="total")
ax.plot(x, nb["N_V_thermal_V2_s"], color="#eda100", label="thermal (Johnson)")
ax.plot(x, nb["N_V_backaction_V2_s"], color="#e87ba4", label="SQUID backaction")
ax.plot(x, nb["N_V_imprecision_V2_s"], color="#2a78d6", label="SQUID imprecision")
corr = nb["N_V_correlation_V2_s"]
ax.plot(x, np.where(corr > 0, corr, np.nan), color="#008300", label="correlation (> 0)")
ax.plot(x, np.where(corr < 0, -corr, np.nan), color="#008300", ls=":", label="|correlation| (< 0)")
ax.set_xscale("symlog", linthresh=10.0, linscale=0.6); ax.set_yscale("log")
ax.set_ylim(nb["N_V_V2_s"].min() / 1e4, nb["N_V_V2_s"].max() * 3)
ax.set_xlabel(r"detuning $x = 2Q_{\rm loaded}(f-f_0)/f_0$ [half-linewidths]")
ax.set_ylabel(r"EMF-referred noise $N_V$ [V$^2$ s]")
ax.set_title(f"toroid + C1+low+RT, f0 = {f0/1e6:g} MHz, T = {T*1e3:.0f} mK, s = +1")
ax.legend(fontsize=7, ncol=3, loc="upper center")
fig.tight_layout()
```

## Scan reach

We distribute three years uniformly in log frequency, adjusting coupling at each tuning. This assumes the required coupling can be realized throughout the band.

We compute and plot scan reach for the amplifier choices to compare their sensitivity.

```python
kw = dict(live_time_yr=3.0, allocation="log_uniform", n=40)
cases = [("ideal amplifier ($\\eta_A=1$)", {"Q": Q, "T": T, "eta_A": 1.0}, (1e5, 3e7), COL["ideal"], "-"),
         ("C1 alone", squid.benchmark("C1", Q, T), (1e5, 3e7), COL["C1"], ":"),
         ("C1+low+RT", squid.benchmark("C1+low+RT", Q, T), (1e5, 3e7), COL["low"], "-"),
         ("C1+high+RT", squid.benchmark("C1+high+RT", Q, T), (1e5, 3e7), COL["high"], "-"),
         ("C1+low+cryo", squid.benchmark("C1+low+cryo", Q, T), (5e6, 3e7), COL["low"], "--")]
reach = {name: scan.reach(PSI, LP, bm, f_min=a, f_max=b, **kw) for name, bm, (a, b), _, _ in cases}
ideal = reach[cases[0][0]]

fig, axes = plt.subplots(1, 2, figsize=(11, 4.2))
for name, bm, _, col, ls in cases:
    r = reach[name]
    axes[0].plot(r["f_Hz"] / 1e6, r["g95_GeV"], color=col, ls=ls, label=name)
    axes[1].plot(r["f_Hz"] / 1e6, r["g95_GeV"] / np.interp(np.log(r["f_Hz"]), np.log(ideal["f_Hz"]), ideal["g95_GeV"]),
                 color=col, ls=ls, label=name)
axes[0].plot(ideal["f_Hz"] / 1e6, ideal["g_target_GeV"], color="0.6", lw=1, label="DFSZ")
axes[0].set_yscale("log"); axes[0].set_ylabel(r"expected 95% limit $g_{a\gamma\gamma}$ [GeV$^{-1}$]")
axes[1].set_ylabel(r"$g_{95}$ / $g_{95}$(ideal amplifier)")
axes[0].set_title(f"toroid R = H = 1 m, B_max = {B_target:g} T, Q = {Q:.0e}, T = {T*1e3:.0f} mK, 3 yr")
axes[1].set_title(r"cost of the real chain ($\propto\bar G^{-1/4}$)")
for ax in axes:
    ax.set_xscale("log"); ax.set_xlabel("frequency [MHz]")
axes[0].legend(fontsize=7)
fig.tight_layout()

for name, *_ in cases:
    r = reach[name]
    print(f"{name.split(' (')[0]:<18} g95 at {r['f_Hz'][0]/1e6:5.2f} MHz = {r['g95_GeV'][0]:.3e} GeV^-1,  eta there = "
          f"{np.atleast_1d(r['band']['eta_A'])[0]:.2f}")
```

Lower axion-coupling limits mean better sensitivity; the right panel shows the penalty from amplifier noise.

## Correlation

We compare noise spectra for `sign=1`, `sign=-1` and zero correlation to assess how correlation affects sensitivity.

```python
variants = {}
for s in (+1, -1):
    ro.squid = squid.chain("C1+low+RT", sign=s).squid_for_resonator(f0, QL, L_eff, T)
    variants[f"s = {s:+d}"] = budget(ro)["N_V_V2_s"]
ro.squid = dataclasses.replace(ro.squid, S_PhiJ=0.0)        # same S_Phi, S_J, correlation dropped
variants["uncorrelated (S_PhiJ = 0)"] = budget(ro)["N_V_V2_s"]

fig, ax = plt.subplots(figsize=(7, 4.2))
for (lab, N), col, ls in zip(variants.items(), ("#2a78d6", "#eb6834", "0.3"), ("-", "--", ":")):
    ax.plot(x, N, color=col, ls=ls, label=lab)
ax.set_xscale("symlog", linthresh=10.0, linscale=0.6); ax.set_yscale("log")
ax.set_xlabel(r"detuning $x$ [half-linewidths]"); ax.set_ylabel(r"$N_V$ [V$^2$ s]")
ax.set_title("C1+low+RT on the toroid: correlation sign and magnitude")
ax.legend(fontsize=8)
fig.tight_layout()

# information per tuning ~ integral of 1/N_V^2 over frequency (signal shape is flat across the band)
for lab, N in variants.items():
    print(f"{lab:<28} min N_V at x = {x[np.argmin(N)]:+7.1f}   int dw/N_V^2 = {np.trapezoid(N**-2, w):.4e}")
```

```python
for s in (+1, -1):
    ro.squid = squid.chain("C1+low+RT", sign=s).squid_for_resonator(f0, QL, L_eff, T)
    e = ro.equivalent(w0)
    print(f"s = {s:+d}: eta_A_eff = {e['eta_A_eff']:.4f}, eta uncorrelated = {e['eta_A_eff_uncorrelated']:.4f}, "
          f"c = {e['correlation_c']:+.3f}, x0 = {e['band_shift_x0']:+.2f}, resid = {e['correlated_reduction_resid']:.1e}")

fe = np.geomspace(1e5, 3e8, 200)
fig, ax = plt.subplots(figsize=(6.4, 4))
for name, col in (("C1", COL["C1"]), ("C1+low+RT", COL["low"]), ("C1+high+RT", COL["high"])):
    ch = squid.chain(name)
    ax.plot(fe / 1e6, ch.eta(fe), color=col, label=f"{name}: correlated $\\eta$")
    ax.plot(fe / 1e6, ch.eta_uncorrelated(fe), color=col, ls=":", label=f"{name}: $\\eta_0$ (correlation dropped)")
ax.set_xscale("log"); ax.set_yscale("log"); ax.set_xlabel("frequency [MHz]"); ax.set_ylabel(r"$\eta$")
ax.set_ylim(1.5, 100)
ax.yaxis.set_major_formatter(plt.ScalarFormatter()); ax.yaxis.set_minor_formatter(plt.NullFormatter())
ax.set_title("the correlation helps most where the first stage dominates")
ax.legend(fontsize=7, ncol=2, loc="lower right")
fig.tight_layout()
```

## Leads, tuning capacitor and the readout network

The previous sections use a lossless loop. This section solves a solenoid pickup, scales its signal coupling $\Psi$ to a peak field of 5 T (the pickup inductance $L_p$ is unchanged by this scaling) and builds the resonant loop with leads and a tuning capacitor. `series_readout` joins the pickup, leads, tuning capacitor and input coil. For this series loop,

$$
L_{\rm eff}=L_p+L_{\rm in}+L_{\rm lead}.
$$

In the classical thermal regime,

$$
\frac{T_{\rm eff}}{Q}=\sum_b\frac{T_b}{Q_b}.
$$

Here $Q$ is the loaded quality factor; $T_b$ is branch temperature and $Q_b$ the quality factor if only branch $b$ dissipated. `equivalent()` returns these effective circuit quantities. The component values below are illustrative.

We solve the solenoid template and assemble the loop.

```python
BLUE, ORANGE, GREEN, GOLD, VIOLET, GREY = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#6f6e69"
PINK, DKGREEN = "#e87ba4", "#008300"

res = af.template("solenoid").solve(RUNS / "solenoid", verbose=0)
print(f"{res.data['mesh']['dofs']} DOFs, warnings: {res.data.get('warnings') or 'none'}")
B_TARGET = 5.0
PSI = res.Psi * B_TARGET / res.B_max
LP = res.L_p
print(f"as solved: Psi = {res.Psi:.4e} H T m^2 at B_max = {res.B_max:.3f} T;  L_p = {LP * 1e9:.1f} nH")
print(f"scaled to B_max = {B_TARGET:g} T: Psi = {PSI:.4e} H T m^2")
```

```python
F0, T, QP = 120.9e3, 0.030, 1e5
NET = dict(Q=QP, T_K=T, L_in_H=0.1 * LP, lead_length_m=0.5, L_per_m=1e-7, R_per_m=1e-9, T_lead_K=4.0, tan_delta=1e-7)
ro = ck.series_readout(L_p_H=LP, Psi_SI=PSI, f_Hz=F0, M_H=1e-9, **NET)     # placeholder SQUID, replaced below
eq0 = ro.equivalent(f_Hz=F0)
print(f"L_eff = {eq0['L_eff_H'] * 1e9:.2f} nH = L_p + L_lead + L_in = {(LP + 0.5e-7 + 0.1 * LP) * 1e9:.2f} nH;  "
      f"participation L_p/L_eff = {eq0['participation']:.4f}")
print(f"R_eff = {eq0['R_eff_ohm']:.4e} ohm, Q_loaded = {eq0['Q_loaded']:.5g}, T_eff = {eq0['T_eff_K'] * 1e3:.2f} mK")

# T_eff/Q = sum_b T_b/Q_b, by hand: Q_b = omega0 L_eff/R_b for each lossy branch
w0 = 2 * math.pi * F0
C_t = 1 / (w0 ** 2 * eq0["L_eff_H"])
R_b = {"pickup (30 mK)": (w0 * LP / QP, T), "leads (4 K)": (0.5 * 1e-9, 4.0), "capacitor (30 mK)": (1e-7 / (w0 * C_t), T)}
TQ = sum(Tb * Rb / (w0 * eq0["L_eff_H"]) for Rb, Tb in R_b.values())
print(f"sum_b T_b/Q_b = {TQ:.5e} K   vs   T_eff/Q_loaded = {eq0['T_eff_K'] / eq0['Q_loaded']:.5e} K")
for k, (Rb, Tb) in R_b.items():
    print(f"   {k:18s} R = {Rb:.3e} ohm, share of T/Q: {Tb * Rb / (w0 * eq0['L_eff_H']) / TQ:6.1%}")
```

The printed noise shares show why warm leads matter even when their resistance is small.

### Attach the SQUID chain

`C1+low+RT` combines a first-stage SQUID, an array and room-temperature preamplifiers. `squid_for_resonator` sets its mutual inductance using the loop's effective inductance, temperature and loaded quality factor.

```python
CHAIN = squid.chain("C1+low+RT")                       # sign=+1; the sign of S_PhiJ is open, see the end
ro.squid = CHAIN.squid_for_resonator(F0, eq0["Q_loaded"], eq0["L_eff_H"], eq0["T_eff_K"])
print(f"eta(f0) = {CHAIN.eta(F0):.3f}, eps_uc/hbar = {CHAIN.eps_uc_over_hbar(F0):.3f}, "
      f"kappa_g^2 = {CHAIN.kappa_g2_scan(F0, eq0['Q_loaded'], eq0['T_eff_K']):.3e}, M = {ro.squid.M_H * 1e12:.1f} pH")
print(ro.summary(F0))
```

`summary()` reports signal flux, noise, sensitivity bandwidth and circuit diagnostics; `ro.draw(f_Hz=F0)` draws the components with their temperatures, as shown for the tuned readout above. The summary uses the effective loop temperature.

### Noise budget

`noise_psd` returns noise referred to the pickup's electromotive force in $\mathrm{V}^2\,\mathrm{s}$. The plot separates thermal noise by branch, backaction, imprecision and the magnitude of their correlation.

```python
eq = ro.equivalent(f_Hz=F0)
bw, lw = eq["bandwidth_Hz"]["noise_doubling"], eq["bandwidth_Hz"]["linewidth"]
fc = eq["band_shift_x0"] * lw / 2                       # band centre, Hz from f0 (x = 2Q(f - f0)/f0)
marks = [fc - bw / 2, fc, fc + bw / 2]
df = np.unique(np.concatenate([-np.geomspace(20 * bw, 1e-3 * bw, 200), [0.0], np.geomspace(1e-3 * bw, 20 * bw, 200), marks]))
keys = ("N_V_V2_s", "N_V_backaction_V2_s", "N_V_imprecision_V2_s", "N_V_correlation_V2_s")
nv = {k: np.empty(df.size) for k in keys}
branches = {}
for i, d in enumerate(df):
    w = 2 * math.pi * (F0 + d)
    p = ro.noise_psd(w)
    for k in keys:
        nv[k][i] = p[k]
    Hs2 = abs(ro.H_signal(w)) ** 2
    for b, v in p["thermal_by_branch_Wb2_s"].items():
        branches.setdefault(b, np.empty(df.size))[i] = v / Hs2

fig, ax = plt.subplots(figsize=(8, 4.6))
ax.plot(df, nv["N_V_V2_s"], color="k", lw=2.2, label="total")
for (b, v), col in zip(branches.items(), (GOLD, ORANGE, GREEN)):
    ax.plot(df, v, color=col, label=f"thermal: {b}")
ax.plot(df, nv["N_V_backaction_V2_s"], color=PINK, label="SQUID backaction")
ax.plot(df, nv["N_V_imprecision_V2_s"], color=BLUE, label="SQUID imprecision")
ax.plot(df, np.abs(nv["N_V_correlation_V2_s"]), color=DKGREEN, ls="--", label="|correlation|")
for xm, ls in zip(marks, (":", "-.", ":")):
    ax.axvline(xm, color=GREY, lw=1, ls=ls)
ax.set_xscale("symlog", linthresh=1.0, linscale=0.5); ax.set_yscale("log")
ax.set_ylim(nv["N_V_V2_s"].min() / 1e5, nv["N_V_V2_s"].max() * 5)
ax.set_xlabel(f"detuning f - f0 [Hz]   (f0 = {F0 / 1e3:.1f} kHz, linewidth {eq['bandwidth_Hz']['linewidth']:.3f} Hz)")
ax.set_ylabel(r"EMF-referred noise $N_V$ [V$^2$ s]")
ax.set_title("solenoid + series readout + C1+low+RT at the optimal coupling", fontsize=9)
ax.legend(fontsize=7.5, ncol=2, loc="upper left")
fig.tight_layout(); plt.show()

idx = {lab: int(np.argmin(np.abs(df - x))) for lab, x in (("f0", 0.0), ("centre", fc), ("lower edge", fc - bw / 2),
                                                          ("upper edge", fc + bw / 2))}
N = nv["N_V_V2_s"]
print(f"band shift x0 = {eq['band_shift_x0']:+.1f} (units of x = 2Q(f-f0)/f0, i.e. half-linewidths) = {fc:+.2f} Hz;  "
      f"noise-doubling bandwidth {bw:.2f} Hz = {bw / lw:.0f} linewidths ({lw:.4f} Hz each)")
print(f"N_V at the band centre = {N[idx['centre']]:.4e} V^2 s; at f0 x{N[idx['f0']] / N[idx['centre']]:.3f}; at the edges "
      f"x{N[idx['lower edge']] / N[idx['centre']]:.3f} and x{N[idx['upper edge']] / N[idx['centre']]:.3f}")
parts = {**{f"thermal: {b}": v for b, v in branches.items()}, "backaction": nv["N_V_backaction_V2_s"],
         "imprecision": nv["N_V_imprecision_V2_s"], "correlation": nv["N_V_correlation_V2_s"]}
print(f"{'share of N_V':24s}" + "".join(f"{lab:>12s}" for lab in idx))
for name, v in parts.items():
    print(f"{name:24s}" + "".join(f"{v[i] / N[i]:12.2%}" for i in idx.values()))
```

Thermal noise and backaction dominate near resonance; imprecision grows away from it. Correlation shifts the noise minimum, marked by the central vertical line; the outer lines mark twice the minimum noise.

## Compare scan sensitivity

Compare an ideal amplifier, a fixed amplifier noise parameter, the SQUID chain alone and the full network over 120.9 kHz–1 MHz. Retune the circuit and recompute its effective inputs at every frequency.

```python
f_band = np.geomspace(120.9e3, 1e6, 15)
cols = {k: [] for k in ("L_eff_H", "Q_loaded", "T_eff_K", "eta_A_eff", "xi_over_opt", "band_shift_x0")}
for f in f_band:
    r = ck.series_readout(L_p_H=LP, Psi_SI=PSI, f_Hz=f, M_H=1e-9, **NET)
    e0 = r.equivalent(f_Hz=f)
    r.squid = CHAIN.squid_for_resonator(f, e0["Q_loaded"], e0["L_eff_H"], e0["T_eff_K"])
    e = r.equivalent(f_Hz=f)
    for k in ("L_eff_H", "Q_loaded", "T_eff_K", "eta_A_eff", "band_shift_x0"):
        cols[k].append(e[k])
    cols["xi_over_opt"].append(e["xi"] / e["xi_opt"])
cols = {k: np.array(v) for k, v in cols.items()}
print(f"over the band: xi/xi_opt = {cols['xi_over_opt'].min():.6f}-{cols['xi_over_opt'].max():.6f}, "
      f"eta_A_eff = {cols['eta_A_eff'].min():.3f}-{cols['eta_A_eff'].max():.3f}, "
      f"T_eff = {cols['T_eff_K'].min() * 1e3:.1f}-{cols['T_eff_K'].max() * 1e3:.1f} mK, "
      f"Q = {cols['Q_loaded'].min():.4g}-{cols['Q_loaded'].max():.4g}, band shift {cols['band_shift_x0'].min():.0f}-"
      f"{cols['band_shift_x0'].max():.0f} (x units, half-linewidths)")

kw = dict(n=60)
cases = {
    "ideal amplifier (eta_A = 1)": scan.reach(PSI, LP, {"Q": QP, "T": T, "eta_A": 1.0}, **kw),
    "conservative eta_A = 10": scan.reach(PSI, LP, {"Q": QP, "T": T, "eta_A": 10.0}, **kw),
    "C1+low+RT, pickup alone": scan.reach(PSI, LP, squid.benchmark(CHAIN, QP, T), **kw),
    "C1+low+RT in the network": scan.reach_dispersive(f_band, PSI, cols["L_eff_H"], Q=cols["Q_loaded"],
                                                      T_K=cols["T_eff_K"], eta_A=cols["eta_A_eff"], **kw),
}
ref = np.median(cases["ideal amplifier (eta_A = 1)"]["g95_GeV"] / cases["ideal amplifier (eta_A = 1)"]["g_target_GeV"])
print(f"\n{'case':30s}{'g95/g_DFSZ':>11s}{'vs ideal':>10s}")
ratio = {}
for name, r in cases.items():
    q = r["g95_GeV"] / r["g_target_GeV"]
    ratio[name] = np.median(q)
    print(f"{name:30s}{ratio[name]:11.3f}{ratio[name] / ref:10.3f}   (flat to {np.ptp(q) / np.median(q):.0e})")

# the thermal-regime estimate of the network step: scan rate ~ Q/(L^2 eta T)  =>  g95 ~ (L^2 eta T/Q)^(1/4)
pred = (cols["L_eff_H"][0] / LP) ** 0.5 * ((cols["T_eff_K"][0] / cols["Q_loaded"][0]) / (T / QP)) ** 0.25
print(f"\nnetwork / pickup alone = {ratio['C1+low+RT in the network'] / ratio['C1+low+RT, pickup alone']:.3f};  "
      f"estimate at 120.9 kHz, (L_eff/L_p)^(1/2) [(T_eff/Q)/(T/Q_p)]^(1/4) = {pred:.3f}")
print(f"conservative / ideal = {ratio['conservative eta_A = 10'] / ref:.3f};  10^(1/4) = {10 ** 0.25:.3f}")

fig, (ax0, ax1) = plt.subplots(1, 2, figsize=(11, 4.2))
styles = {"ideal amplifier (eta_A = 1)": ("0.15", "-"), "conservative eta_A = 10": (ORANGE, "--"),
          "C1+low+RT, pickup alone": (BLUE, "-"), "C1+low+RT in the network": (VIOLET, "-")}
r0 = cases["ideal amplifier (eta_A = 1)"]
ax0.loglog(r0["m_eV"] * 1e9, r0["g_target_GeV"], color=GREY, lw=1, label="DFSZ")
for name, r in cases.items():
    ax0.loglog(r["m_eV"] * 1e9, r["g95_GeV"], color=styles[name][0], ls=styles[name][1], label=name)
ax0.set_xticks([0.5, 1, 2, 4]); ax0.set_xticklabels(["0.5", "1", "2", "4"])
ax0.xaxis.set_minor_formatter(plt.NullFormatter())
ax0.set_xlabel("$m_a$ [neV]"); ax0.set_ylabel(r"expected 95% limit $g_{a\gamma\gamma}$ [GeV$^{-1}$]")
ax0.set_title(f"solenoid template at B_max = {B_TARGET:g} T, 3 yr, target allocation", fontsize=9); ax0.legend(fontsize=7.5)
fk = f_band / 1e3
ax1.semilogx(fk, CHAIN.eta(f_band), color=BLUE, label=r"chain $\eta(f)$")
ax1.semilogx(fk, cols["T_eff_K"] / T, color=GOLD, label=r"$T_{\rm eff}/T$ (warm leads)")
ax1.semilogx(fk, cols["L_eff_H"] / LP, color=VIOLET, label=r"$L_{\rm eff}/L_p$")
ax1.semilogx(fk, cols["Q_loaded"] / QP, color=GREEN, label=r"$Q_{\rm loaded}/Q_p$")
ax1.axhline(1, color=GREY, lw=0.8)
ax1.set_xlabel("tuning frequency [kHz]"); ax1.set_ylabel("value"); ax1.set_yscale("log")
ax1.set_yticks([1, 2, 5, 10]); ax1.set_yticklabels(["1", "2", "5", "10"]); ax1.yaxis.set_minor_formatter(plt.NullFormatter())
ax1.set_xticks([125, 250, 500, 1000]); ax1.set_xticklabels(["125", "250", "500", "1000"])
ax1.xaxis.set_minor_formatter(plt.NullFormatter())
ax1.set_title("the reduced scan inputs of the network, per tuning", fontsize=9); ax1.legend(fontsize=8)
fig.tight_layout(); plt.show()
```

Lower $g_{95}/g_{\rm DFSZ}$ means better sensitivity; values below one reach the DFSZ target. Extra inductance dilutes the signal, and warm losses increase the effective temperature.

The correlation sign is unspecified. The next cell compares both choices; for this series loop at high quality factor, they shift the sensitive band in opposite directions without changing the leading scan rate.

```python
for s in (+1, -1):
    r = ck.series_readout(L_p_H=LP, Psi_SI=PSI, f_Hz=F0, M_H=1e-9, **NET)
    r.squid = squid.chain("C1+low+RT", sign=s).squid_for_resonator(F0, eq0["Q_loaded"], eq0["L_eff_H"], eq0["T_eff_K"])
    e = r.equivalent(f_Hz=F0)
    print(f"s = {s:+d}: eta_A_eff = {e['eta_A_eff']:.4f} (uncorrelated {e['eta_A_eff_uncorrelated']:.4f}), "
          f"band shift x0 = {e['band_shift_x0']:+.1f} half-linewidths, G_bar = {e['G_bar']:.5e}")
```

The second stage is scaled from a measured prototype; the chain model neglects SQUID input loading and uses a white-noise approximation.
